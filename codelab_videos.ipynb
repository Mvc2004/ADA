```python
import cv2, dlib, time

detector = dlib.get_frontal_face_detector()
cap = cv2.VideoCapture('video_rostros.mp4')  # tu video

prev = time.time()
frames = 0

while True:
    ok, frame = cap.read()
    if not ok:
        break

    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    rects = detector(gray, 0)  # detector HOG de dlib

    for r in rects:
        x, y, w, h = r.left(), r.top(), r.width(), r.height()
        cv2.rectangle(frame, (x,y), (x+w,y+h), (0,255,0), 2)

    frames += 1

    if frames % 10 == 0:
        now = time.time()
        fps = 10/(now-prev)
        prev = now

        cv2.putText(frame, f"FPS: {fps:.1f}", (10,30),
                    cv2.FONT_HERSHEY_SIMPLEX, 1, (0,255,0), 2)

cap.release()
```

```python
import cv2
import dlib
import time
import numpy as np

# Detector de rostros HOG de dlib
detector = dlib.get_frontal_face_detector()

# Abrir el video
cap = cv2.VideoCapture('video_rostros.mp4')

frames = 0
fps_list = []

while True:

    start = time.time()

    ok, frame = cap.read()

    if not ok:
        break

    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    # 0 = sin upsampling
    rects = detector(gray, 0)

    for r in rects:
        x = r.left()
        y = r.top()
        w = r.width()
        h = r.height()

        cv2.rectangle(frame, (x,y), (x+w,y+h), (0,255,0), 2)

    end = time.time()

    fps = 1/(end-start)

    fps_list.append(fps)

    frames += 1

    # Guardar un frame con detecciones
    if frames == 50:
        cv2.imwrite("frame_detectado.jpg", frame)

    # Solo medir 200 frames
    if frames >= 200:
        break


cap.release()

print("Frames procesados:", frames)
print("FPS promedio:", np.mean(fps_list))
```
