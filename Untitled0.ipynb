```python
def iou(a, b):
    # a,b en formato [x,y,w,h]
    ax1, ay1, aw, ah = a; ax2, ay2 = ax1+aw, ay1+ah
    bx1, by1, bw, bh = b; bx2, by2 = bx1+bw, by1+bh
    ix1, iy1 = max(ax1,bx1), max(ay1,by1)
    ix2, iy2 = min(ax2,bx2), min(ay2,by2)
    inter = max(0, ix2-ix1)*max(0, iy2-iy1)
    union = aw*ah + bw*bh - inter
    return inter/union if union>0 else 0.0
```

```python
import cv2
import matplotlib.pyplot as plt
from mtcnn.mtcnn import MTCNN
from time import time

detector = MTCNN()

def detectar_rostros(nombre_imagen):

    img = cv2.imread(nombre_imagen)
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    t0 = time()
    res = detector.detect_faces(img_rgb)
    t1 = time()

    print(f"Imagen: {nombre_imagen}")
    print(f"Rostros detectados: {len(res)}")
    print(f"Tiempo: {(t1-t0)*1000:.2f} ms\n")

    return res, img_rgb
```

```python
res1, img1 = detectar_rostros("persona.jpg")
res2, img2 = detectar_rostros("grupo.jpg")
res3, img3 = detectar_rostros("parcial.jpg")
```

```python
import cv2
import matplotlib.pyplot as plt
from mtcnn.mtcnn import MTCNN
from time import time

detector = MTCNN()

imagenes = ["persona.jpg", "grupo.jpg", "parcial.jpg"]

plt.figure(figsize=(15,5))

for i, nombre in enumerate(imagenes):

    img = cv2.imread(nombre)
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    t0 = time()
    res = detector.detect_faces(img_rgb)
    t1 = time()

    print(f"{nombre} → {len(res)} rostros detectados")
    print(f"Tiempo: {(t1-t0)*1000:.2f} ms\n")

    for r in res:

        x,y,w,h = r['box']
        cv2.rectangle(img_rgb,(x,y),(x+w,y+h),(0,255,0),2)

        for punto in r['keypoints'].values():
            cv2.circle(img_rgb,punto,3,(255,0,0),-1)

    plt.subplot(1,3,i+1)
    plt.imshow(img_rgb)
    plt.title(nombre)
    plt.axis("off")

plt.show()
```
